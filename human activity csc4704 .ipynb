## Prepare data for visualization

### Subtask:
Merge feature names to the X dataframes and activity labels to the y dataframes. Combine subject information. Specifically, this involves renaming columns, concatenating training and test sets for X, y, and subject, and then merging them into a single, comprehensive DataFrame for easier analysis.


**Reasoning**:
I will perform the data preparation steps as outlined in the instructions. This involves extracting feature names and assigning them, renaming columns for activity IDs and subject IDs, concatenating training and test datasets for features, activities, and subjects, combining them into a single DataFrame, and finally merging with activity labels to create `df_final`. I will then display the first few rows and the shape of the final DataFrame for verification.



```python
import pandas as pd

# 1. Extract feature names and assign them as column names to X_train_df and X_test_df
feature_names = features_df.iloc[:, 1].tolist()
X_train_df.columns = feature_names
X_test_df.columns = feature_names

# 2. Rename the single column of y_train_df and y_test_df to 'activity_id'
y_train_df.columns = ['activity_id']
y_test_df.columns = ['activity_id']

# 3. Rename the single column of subject_train_df and subject_test_df to 'subject_id'
subject_train_df.columns = ['subject_id']
subject_test_df.columns = ['subject_id']

# 4. Concatenate X_train_df and X_test_df vertically
X_combined_df = pd.concat([X_train_df, X_test_df], axis=0, ignore_index=True)
print(f"Shape of X_combined_df: {X_combined_df.shape}")

# 5. Concatenate y_train_df and y_test_df vertically
y_combined_df = pd.concat([y_train_df, y_test_df], axis=0, ignore_index=True)
print(f"Shape of y_combined_df: {y_combined_df.shape}")

# 6. Concatenate subject_train_df and subject_test_df vertically
subject_combined_df = pd.concat([subject_train_df, subject_test_df], axis=0, ignore_index=True)
print(f"Shape of subject_combined_df: {subject_combined_df.shape}")

# 7. Concatenate X_combined_df, y_combined_df, and subject_combined_df horizontally
df_combined = pd.concat([X_combined_df, y_combined_df, subject_combined_df], axis=1)
print(f"Shape of df_combined: {df_combined.shape}")

# 8. Merge df_combined with activity_labels_df based on the 'activity_id' column
df_final = pd.merge(df_combined, activity_labels_df, on='activity_id', how='left')
print(f"Shape of df_final: {df_final.shape}")

# 9. Display the first 5 rows and the shape of df_final
print("\nFirst 5 rows of df_final:")
print(df_final.head())
```

## Visualize Activity Distribution

### Subtask:
Create a bar chart showing the distribution of each activity in the combined dataset. This will help understand the balance of activities.


**Reasoning**:
I will create a bar chart to visualize the distribution of activities in the `df_final` DataFrame, as requested by the subtask. This involves importing visualization libraries, using `seaborn.countplot` to plot activity counts, and adding labels and a title for clarity.



```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set the figure size for better readability
plt.figure(figsize=(10, 6))

# Create a bar plot of the activity_name column
sns.countplot(x='activity_name', data=df_final, palette='viridis')

# Add title and labels
plt.title('Distribution of Activities', fontsize=16)
plt.xlabel('Activity', fontsize=12)
plt.ylabel('Count', fontsize=12)

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Adjust layout to prevent labels from overlapping
plt.tight_layout()

# Display the plot
plt.show()
```

**Reasoning**:
The previous code generated a `FutureWarning` because `palette` was used without assigning a `hue` in `sns.countplot`. I will update the `sns.countplot` call to assign `x` to `hue` and set `legend=False` as suggested by the warning.



```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set the figure size for better readability
plt.figure(figsize=(10, 6))

# Create a bar plot of the activity_name column
sns.countplot(x='activity_name', data=df_final, palette='viridis', hue='activity_name', legend=False)

# Add title and labels
plt.title('Distribution of Activities', fontsize=16)
plt.xlabel('Activity', fontsize=12)
plt.ylabel('Count', fontsize=12)

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Adjust layout to prevent labels from overlapping
plt.tight_layout()

# Display the plot
plt.show()
```
