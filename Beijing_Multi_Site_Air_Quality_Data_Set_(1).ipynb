## Load and Unzip Data

### Subtask:
Unzip the provided /content/archive (6).zip file and load the relevant dataset(s) into a pandas DataFrame(s). If there are multiple files after unzipping, merge them as necessary based on the project context (e.g., if they represent different stations or time periods).


**Reasoning**:
First, I'll import necessary libraries, create a directory for the unzipped files, and then extract the contents of the provided zip file into that directory. This sets up the environment for further data loading.



```python
import zipfile
import os
import pandas as pd

# Define the path to the zip file and the extraction directory
zip_file_path = '/content/archive (6).zip'
extraction_dir = 'unzipped_data'

# Create the extraction directory if it doesn't exist
os.makedirs(extraction_dir, exist_ok=True)

# Unzip the file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extraction_dir)

print(f"'{zip_file_path}' unzipped to '{extraction_dir}' successfully.")

# List the contents of the extracted directory to identify the primary dataset file(s)
extracted_files = os.listdir(extraction_dir)
print("Contents of the extracted directory:")
for file_name in extracted_files:
    print(f"- {file_name}")
```

**Reasoning**:
Now that the files are unzipped and listed, I will load each CSV file into a pandas DataFrame and store them in a dictionary. This will allow for inspection of each dataset before deciding on a merging strategy.



```python
dataframes = {}
for file_name in extracted_files:
    if file_name.endswith('.csv'):
        file_path = os.path.join(extraction_dir, file_name)
        # Extract station name from the filename for better identification
        station_name = file_name.replace('PRSA_Data_', '').replace('_20130301-20170228.csv', '')
        df = pd.read_csv(file_path)
        dataframes[station_name] = df
        print(f"Loaded '{file_name}' into DataFrame '{station_name}'. Shape: {df.shape}")

# Display the first few rows and info of one of the dataframes to understand its structure
if dataframes:
    first_station_name = list(dataframes.keys())[0]
    print(f"\nFirst 5 rows of '{first_station_name}' dataset:")
    print(dataframes[first_station_name].head())
    print(f"\nInfo for '{first_station_name}' dataset:")
    dataframes[first_station_name].info()
```

## Data Cleaning

### Subtask:
Handle missing values, identify and address outliers/anomalies, and correct data types to prepare the dataset for modeling.


**Reasoning**:
First, I need to address the anomalous values in the specified pollutant columns by replacing them with NaN, as these values likely represent placeholders for missing or extreme data rather than actual measurements. This prepares these columns for proper imputation.



```python
import numpy as np

# Convert anomalous values in pollutant columns to NaN
all_data['PM2.5'] = all_data['PM2.5'].replace(999.0, np.nan)
all_data['PM10'] = all_data['PM10'].replace(999.0, np.nan)
all_data['CO'] = all_data['CO'].replace(10000.0, np.nan)
all_data['O3'] = all_data['O3'].replace(1071.0, np.nan)

print("Anomalous values in PM2.5, PM10, CO, and O3 converted to NaN.")

# Display the count of NaN values after replacement for these columns
print("\nNaN counts after anomalous value replacement:")
print(all_data[['PM2.5', 'PM10', 'CO', 'O3']].isnull().sum())
```

**Reasoning**:
Following the instructions, I will now fill missing numerical values in the `all_data` DataFrame using a two-step imputation process per 'station': first, forward-fill with a limit, then backward-fill with a similar limit, and finally linear interpolation for any remaining gaps.



```python
numerical_cols_to_impute = [
    'PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3',
    'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM'
]

for col in numerical_cols_to_impute:
    # Step 1: Forward-fill with a limit of 3 hours within each station group
    all_data[col] = all_data.groupby('station')[col].ffill(limit=3)

    # Step 2: Backward-fill with a limit of 3 hours within each station group
    all_data[col] = all_data.groupby('station')[col].bfill(limit=3)

    # Step 3: Linear interpolation for any remaining gaps, also per station
    all_data[col] = all_data.groupby('station')[col].transform(lambda group: group.interpolate(method='linear', limit_direction='both'))

print("Missing numerical values imputed using ffill, bfill, and linear interpolation per station.")

# Display the count of NaN values after imputation for these columns
print("\nNaN counts after numerical imputation:")
print(all_data[numerical_cols_to_impute].isnull().sum().sort_values(ascending=False))
```

**Reasoning**:
Now that numerical columns are handled, I will fill missing values in the categorical 'wd' (wind direction) column using forward-fill and then backward-fill, grouped by 'station', as it is a time-series categorical feature.



```python
all_data['wd'] = all_data.groupby('station')['wd'].ffill()
all_data['wd'] = all_data.groupby('station')['wd'].bfill()

print("Missing values in 'wd' column imputed using ffill and bfill per station.")

# Display the count of NaN values in 'wd' after imputation
print("\nNaN count in 'wd' column after imputation:")
print(all_data['wd'].isnull().sum())
```

## Feature Engineering

### Subtask:
Create new time-based features from the 'datetime' column and encode categorical variables.


**Reasoning**:
I will extract various time-based features (hour, day of week, day of year, week of year, month, quarter) from the 'datetime' column and create a new 'season' column based on the month, as instructed.



```python
all_data['hour'] = all_data['datetime'].dt.hour
all_data['day_of_week'] = all_data['datetime'].dt.dayofweek
all_data['day_of_year'] = all_data['datetime'].dt.dayofyear
all_data['week_of_year'] = all_data['datetime'].dt.isocalendar().week.astype(int)
all_data['month'] = all_data['datetime'].dt.month
all_data['quarter'] = all_data['datetime'].dt.quarter

def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Autumn'

all_data['season'] = all_data['month'].apply(get_season)

print("Time-based features and 'season' column created.")
print("First 5 rows with new time-based features:")
print(all_data[['datetime', 'hour', 'day_of_week', 'day_of_year', 'week_of_year', 'month', 'quarter', 'season']].head())
```

**Reasoning**:
Now that the time-based features are created, I will proceed with one-hot encoding the categorical 'wd' and 'station' columns to convert them into numerical representations suitable for modeling. After encoding, I will display the first 5 rows and the info of the DataFrame.



```python
all_data = pd.get_dummies(all_data, columns=['wd'], prefix='wd')
all_data = pd.get_dummies(all_data, columns=['station'], prefix='station')

print("Categorical 'wd' and 'station' columns have been one-hot encoded.")

print("\nFirst 5 rows of the DataFrame after encoding:")
print(all_data.head())
print("\nInfo for the DataFrame after encoding:")
all_data.info()
```
